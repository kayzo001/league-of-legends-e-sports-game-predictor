```python
# 1. IMPORTS & SETUP
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import joblib
import os
import warnings

from IPython.display import display
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier, AdaBoostClassifier
from sklearn.svm import SVC
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, f1_score

# Notebook configuration
%matplotlib inline
sns.set_theme(style="whitegrid")
warnings.filterwarnings('ignore')

print("Environment initialized.")
```

```python
# 2. DATA LOADING
print("Loading Oracle's Elixir dataset...")
file_path = "2024_LoL_esports_match_data_from_OraclesElixir1.csv"

# Keep the raw dataframe unchanged so EDA can use all original columns.
raw_df = pd.read_csv(file_path, sep=';', low_memory=False)

# Create a separate modeling dataframe where unused ID/meta columns can be removed.
useless_columns = ['url', 'league', 'year', 'split', 'date', 'patch', 'playerid', 'teamid']
model_df = raw_df.drop(columns=[col for col in useless_columns if col in raw_df.columns]).copy()

# Keep df as a simple alias for compatibility with the original notebook style.
df = model_df.copy()

print(f"Raw dataset shape: {raw_df.shape[0]} rows and {raw_df.shape[1]} columns")
print(f"Modeling dataset shape after light pruning: {df.shape[0]} rows and {df.shape[1]} columns")
```

```python
# 3. INITIAL EDA ON RAW DATA
print("Initial dataset overview")
print(f"Raw shape: {raw_df.shape}")
print(f"Column count: {raw_df.shape[1]}")

display(raw_df.head())

print("\nBasic dataframe info:")
raw_df.info()

# Show game and row structure before any preprocessing.
if 'gameid' in raw_df.columns:
    print(f"\nNumber of unique games: {raw_df['gameid'].nunique()}")
else:
    print("\nColumn 'gameid' is missing, so unique games cannot be counted.")

if 'position' in raw_df.columns:
    position_counts = raw_df['position'].value_counts(dropna=False)
    print("\nRows by position:")
    display(position_counts.to_frame(name='row_count'))
    raw_team_rows = raw_df[raw_df['position'] == 'team'].copy()
else:
    print("\nColumn 'position' is missing, so team rows cannot be separated yet.")
    raw_team_rows = raw_df.copy()

print(f"Number of team rows: {len(raw_team_rows)}")

# Target analysis is done on team rows because each match has two team-level outcomes.
if 'result' in raw_team_rows.columns:
    raw_target = pd.to_numeric(raw_team_rows['result'], errors='coerce').dropna()
    target_counts = raw_target.value_counts().sort_index()
    target_percentages = (raw_target.value_counts(normalize=True).sort_index() * 100).round(2)
    target_summary = pd.DataFrame({
        'count': target_counts,
        'percentage': target_percentages
    })
    display(target_summary)

    plt.figure(figsize=(5, 4))
    sns.countplot(data=pd.DataFrame({'result': raw_target.astype(int)}), x='result', palette=['#d95f5f', '#4c9f70'])
    plt.title('Team Result Counts')
    plt.xlabel('Result (0 = Loss, 1 = Win)')
    plt.ylabel('Number of Team Rows')
    plt.show()

    if len(target_percentages) == 2 and abs(target_percentages.iloc[0] - target_percentages.iloc[1]) <= 10:
        print("Interpretation: the target classes are fairly balanced.")
    else:
        print("Interpretation: the target classes may be imbalanced, so model metrics should be checked carefully.")
else:
    print("Column 'result' is missing, so target analysis cannot be shown.")

# Missing value table for the full raw dataset.
missing_table = pd.DataFrame({
    'column': raw_df.columns,
    'missing_count': raw_df.isna().sum().values,
    'missing_percentage': (raw_df.isna().mean().values * 100).round(2)
})
missing_table = missing_table[missing_table['missing_count'] > 0].sort_values('missing_percentage', ascending=False)

print("\nColumns with missing values:")
if missing_table.empty:
    print("No missing values found in the raw dataset.")
else:
    display(missing_table)

# Specific missing value check for the columns used later in the project.
feature_candidates = [
    'golddiffat15', 'xpdiffat15', 'turretplates', 'opp_turretplates',
    'killsat15', 'assistsat15', 'deathsat15', 'side', 'result'
]
feature_missing_rows = []
for col in feature_candidates:
    if col in raw_df.columns:
        feature_missing_rows.append({
            'column': col,
            'available': True,
            'missing_count': raw_df[col].isna().sum(),
            'missing_percentage': round(raw_df[col].isna().mean() * 100, 2)
        })
    else:
        feature_missing_rows.append({
            'column': col,
            'available': False,
            'missing_count': np.nan,
            'missing_percentage': np.nan
        })

print("\nMissing values in final feature candidates:")
display(pd.DataFrame(feature_missing_rows))

# Basic numeric summaries for important 15-minute columns.
important_15_cols = [
    'golddiffat15', 'xpdiffat15', 'turretplates', 'opp_turretplates',
    'killsat15', 'assistsat15', 'deathsat15'
]
available_15_cols = [col for col in important_15_cols if col in raw_df.columns]

if available_15_cols:
    numeric_summary = raw_df[available_15_cols].apply(pd.to_numeric, errors='coerce').describe().T
    print("\nDescriptive statistics for important 15-minute columns:")
    display(numeric_summary)
else:
    print("\nNo important 15-minute columns were found for numeric summaries.")
```

```python
# 4. TEAM-LEVEL FILTERING AND FEATURE ENGINEERING
print("Engineering 15-minute team-state features...")

# Keep only team rows for the model target and team-level game-state features.
if 'position' in df.columns:
    teams = df[df['position'] == 'team'].copy()
else:
    teams = df.copy()
    print("Column 'position' is missing; using all rows as a fallback.")

# Make the required columns exist so later code fails gracefully instead of raising KeyError.
required_team_columns = ['gameid', 'side', 'result']
numeric_cols = [
    'golddiffat15', 'xpdiffat15', 'turretplates', 'opp_turretplates',
    'killsat15', 'assistsat15', 'deathsat15', 'result'
]
for col in required_team_columns + numeric_cols:
    if col not in teams.columns:
        teams[col] = np.nan

# Convert important numeric columns before creating new features.
for col in numeric_cols:
    teams[col] = pd.to_numeric(teams[col], errors='coerce')

# Rows without the target, game ID, or side cannot be used safely.
start_rows = len(teams)
teams = teams.dropna(subset=['result', 'gameid', 'side']).copy()
print(f"Dropped {start_rows - len(teams)} team rows missing result, gameid, or side.")

# Keep target as a clean 0/1 value.
teams['result'] = teams['result'].astype(int)

# Feature 1: plate difference at 15 minutes.
teams['platediff'] = teams['turretplates'] - teams['opp_turretplates']

# Feature 2: simple 15-minute KDA. Add 1 to deaths to avoid division by zero.
teams['kda_15'] = (teams['killsat15'] + teams['assistsat15']) / (teams['deathsat15'] + 1)
teams['kda_15'] = teams['kda_15'].replace([np.inf, -np.inf], np.nan)

# Feature 3: encode side in the same way as the original prototype.
teams['side_encoded'] = teams['side'].map({'Blue': 1, 'Red': 0})

print(f"Engineered base features for {teams.shape[0]} team rows across {teams['gameid'].nunique()} games.")
display(teams[['gameid', 'side', 'result', 'golddiffat15', 'xpdiffat15', 'platediff', 'kda_15', 'side_encoded']].head())
```

```python
# 5. TRAIN/TEST SPLIT BY GAME ID
print("Splitting by unique game IDs to avoid leakage between teams from the same game...")

unique_game_ids = pd.Series(teams['gameid'].dropna().unique())

if len(unique_game_ids) < 2:
    raise ValueError("Not enough unique games are available for a train/test split.")

# A simple game-level stratification label: did Blue side win this game?
# If this is not available or not suitable, we fall back to a normal split.
stratify_labels = None
if {'gameid', 'side', 'result'}.issubset(teams.columns):
    blue_results = teams[teams['side'] == 'Blue'].drop_duplicates('gameid').set_index('gameid')['result']
    possible_labels = unique_game_ids.map(blue_results)
    label_counts = possible_labels.value_counts(dropna=True)
    if possible_labels.notna().all() and possible_labels.nunique() > 1 and label_counts.min() >= 2:
        stratify_labels = possible_labels
        print("Using Blue-side game result for stratified game-level splitting.")
    else:
        print("Using a normal game-level split because stratification is not suitable.")

train_game_ids, test_game_ids = train_test_split(
    unique_game_ids,
    test_size=0.2,
    random_state=42,
    stratify=stratify_labels
)

train_teams = teams[teams['gameid'].isin(train_game_ids)].copy()
test_teams = teams[teams['gameid'].isin(test_game_ids)].copy()

print(f"Training games: {len(train_game_ids)}")
print(f"Testing games: {len(test_game_ids)}")
print(f"Training team rows: {train_teams.shape[0]}")
print(f"Testing team rows: {test_teams.shape[0]}")
```

```python
# 6. DRAFT ADVANTAGE ENCODING WITHOUT DATA LEAKAGE
print("Creating smoothed champion win rates from training player rows only...")

# Player rows are needed only for champion draft information.
if 'position' in df.columns:
    players_df = df[df['position'] != 'team'].copy()
else:
    players_df = pd.DataFrame(columns=df.columns)

for col in ['gameid', 'side', 'position', 'champion', 'result']:
    if col not in players_df.columns:
        players_df[col] = np.nan

players_df['result'] = pd.to_numeric(players_df['result'], errors='coerce')
train_players = players_df[players_df['gameid'].isin(train_game_ids)].copy()

# Smoothing prevents rare champions from receiving extreme win rates.
smoothing = 10
global_champ_win_rate = train_players['result'].mean()
if pd.isna(global_champ_win_rate):
    global_champ_win_rate = 0.5

champ_stats = (
    train_players.dropna(subset=['champion', 'result'])
    .groupby('champion')['result']
    .agg(champion_wins='sum', champion_games='count')
)
champ_stats['smoothed_winrate'] = (
    champ_stats['champion_wins'] + smoothing * global_champ_win_rate
) / (champ_stats['champion_games'] + smoothing)

train_champ_win_rates = champ_stats['smoothed_winrate'].to_dict()

print(f"Global training champion win rate: {global_champ_win_rate:.3f}")
print(f"Champion win rates learned for {len(train_champ_win_rates)} champions.")
display(champ_stats.sort_values('champion_games', ascending=False).head())


def engineer_draft(team_subset, player_subset, win_rates, global_win_rate=0.5):
    """Create draft_adv using only win rates learned from the training split."""
    result_df = team_subset.copy()

    if result_df.empty or player_subset.empty:
        result_df['draft_adv'] = 0
        return result_df

    required_player_cols = ['gameid', 'side', 'position', 'champion']
    if any(col not in player_subset.columns for col in required_player_cols):
        result_df['draft_adv'] = 0
        return result_df

    game_ids = result_df['gameid'].dropna().unique()
    p_sub = player_subset[player_subset['gameid'].isin(game_ids)].copy()

    if p_sub.empty:
        result_df['draft_adv'] = 0
        return result_df

    p_sub['champ_winrate'] = p_sub['champion'].map(win_rates).fillna(global_win_rate)

    expected_positions = ['top', 'jng', 'mid', 'bot', 'sup']
    draft_feats = (
        p_sub.pivot_table(
            index=['gameid', 'side'],
            columns='position',
            values='champ_winrate',
            aggfunc='mean'
        )
        .reset_index()
    )

    # Average only positions that exist in this subset, so missing positions do not break the notebook.
    available_positions = [pos for pos in expected_positions if pos in draft_feats.columns]
    if available_positions:
        draft_feats['team_draft_strength'] = draft_feats[available_positions].mean(axis=1).fillna(global_win_rate)
    else:
        draft_feats['team_draft_strength'] = global_win_rate

    strengths = draft_feats[['gameid', 'side', 'team_draft_strength']].copy()

    # Match each team with the opposite side from the same game.
    opp_draft = strengths.copy()
    opp_draft['side'] = opp_draft['side'].map({'Blue': 'Red', 'Red': 'Blue'})
    opp_draft = opp_draft.rename(columns={'team_draft_strength': 'opp_draft_strength'})

    final_draft = strengths.merge(opp_draft, on=['gameid', 'side'], how='left')
    final_draft['draft_adv'] = (
        final_draft['team_draft_strength'] - final_draft['opp_draft_strength']
    ).fillna(0)

    result_df = result_df.merge(final_draft[['gameid', 'side', 'draft_adv']], on=['gameid', 'side'], how='left')
    result_df['draft_adv'] = result_df['draft_adv'].fillna(0)

    return result_df


train_master = engineer_draft(train_teams, players_df, train_champ_win_rates, global_champ_win_rate)
test_master = engineer_draft(test_teams, players_df, train_champ_win_rates, global_champ_win_rate)

print("Draft advantage created without using test results.")
display(train_master[['gameid', 'side', 'draft_adv']].head())
```

```python
# 7. IMPROVED EDA ON ENGINEERED TRAINING DATA
features = ['golddiffat15', 'xpdiffat15', 'platediff', 'kda_15', 'draft_adv', 'side_encoded']
dist_features = ['golddiffat15', 'xpdiffat15', 'platediff', 'kda_15', 'draft_adv']
result_palette = ['#d95f5f', '#4c9f70']

eda_train = train_master.copy()
for col in features + ['result']:
    if col not in eda_train.columns:
        eda_train[col] = np.nan
    eda_train[col] = pd.to_numeric(eda_train[col], errors='coerce')

print("Feature distributions by result on the training set")
for feature in dist_features:
    if eda_train[feature].notna().any():
        plt.figure(figsize=(7, 4))
        sns.histplot(
            data=eda_train,
            x=feature,
            hue='result',
            bins=30,
            kde=True,
            stat='density',
            common_norm=False,
            palette=result_palette
        )
        plt.title(f'{feature} Distribution by Result')
        plt.xlabel(feature)
        plt.ylabel('Density')
        plt.show()
    else:
        print(f"Skipping {feature}: no numeric values available.")

print("Boxplots by result")
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
axes = axes.flatten()
for ax, feature in zip(axes, dist_features):
    if eda_train[feature].notna().any():
        sns.boxplot(data=eda_train, x='result', y=feature, ax=ax, palette=result_palette)
        ax.set_title(f'{feature} by Result')
        ax.set_xlabel('Result (0 = Loss, 1 = Win)')
    else:
        ax.set_visible(False)

for ax in axes[len(dist_features):]:
    ax.set_visible(False)

plt.tight_layout()
plt.show()

# Side analysis
if 'side' in eda_train.columns and eda_train['result'].notna().any():
    side_win_rate = eda_train.groupby('side', dropna=False)['result'].mean().reset_index()
    side_win_rate['win_rate_percentage'] = (side_win_rate['result'] * 100).round(2)
    display(side_win_rate.rename(columns={'result': 'win_rate'}))

    plt.figure(figsize=(5, 4))
    sns.barplot(data=side_win_rate, x='side', y='result', palette=['#4c78a8', '#d95f5f'])
    plt.title('Win Rate by Side')
    plt.xlabel('Side')
    plt.ylabel('Win Rate')
    plt.ylim(0, 1)
    plt.show()

    best_side = side_win_rate.sort_values('result', ascending=False).iloc[0]
    print(f"Interpretation: {best_side['side']} has the higher observed win rate in the training data ({best_side['result']:.2%}).")
else:
    print("Side analysis skipped because side or result is missing.")

# Correlation heatmap using numeric training columns only.
corr_source = eda_train[features + ['result']].select_dtypes(include=[np.number])
if 'result' in corr_source.columns and corr_source.shape[1] > 1:
    corr_matrix = corr_source.corr()
    plt.figure(figsize=(10, 6))
    sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', vmin=-1, vmax=1, fmt='.2f')
    plt.title('Correlation Matrix on Training Data')
    plt.show()

    result_corr = corr_matrix['result'].drop('result').dropna()
    result_corr = result_corr.reindex(result_corr.abs().sort_values(ascending=False).index)
    print("Most correlated features with result:")
    display(result_corr.to_frame(name='correlation_with_result'))

    if not result_corr.empty:
        top_feature = result_corr.index[0]
        print(f"Interpretation: {top_feature} has the strongest linear relationship with result among these features.")
else:
    print("Correlation heatmap skipped because numeric columns are missing.")

# Simple outlier check. We display extreme values but do not remove them automatically.
print("Outlier summary for final features:")
outlier_summary = eda_train[features].agg(['min', 'max', 'mean', 'median']).T
outlier_summary = outlier_summary.round(3)
display(outlier_summary)
print("These min and max values should be inspected, but no outliers are removed automatically in this prototype.")
```

```python
# 8. MISSING VALUE HANDLING AND FINAL PREPROCESSING
print("Creating final clean feature matrices...")

features = ['golddiffat15', 'xpdiffat15', 'platediff', 'kda_15', 'draft_adv', 'side_encoded']
median_features = ['golddiffat15', 'xpdiffat15', 'platediff', 'kda_15']

# Ensure expected columns exist and are numeric in both splits.
for frame in [train_master, test_master]:
    for col in features + ['result']:
        if col not in frame.columns:
            frame[col] = np.nan
        frame[col] = pd.to_numeric(frame[col], errors='coerce')

# Drop rows with missing targets after all numeric conversion.
train_master = train_master.dropna(subset=['result']).copy()
test_master = test_master.dropna(subset=['result']).copy()

# Fit imputation values on training data only.
median_fill_values = train_master[median_features].median().fillna(0)
side_mode = train_master['side_encoded'].mode(dropna=True)
side_fill_value = side_mode.iloc[0] if not side_mode.empty else 0

# Apply the training-set imputation values to train and test.
for col in median_features:
    train_master[col] = train_master[col].fillna(median_fill_values[col])
    test_master[col] = test_master[col].fillna(median_fill_values[col])

train_master['draft_adv'] = train_master['draft_adv'].fillna(0)
test_master['draft_adv'] = test_master['draft_adv'].fillna(0)

train_master['side_encoded'] = train_master['side_encoded'].fillna(side_fill_value)
test_master['side_encoded'] = test_master['side_encoded'].fillna(side_fill_value)

# Build final model matrices and keep them as DataFrames for readability.
X_train = train_master[features].apply(pd.to_numeric, errors='coerce')
X_test = test_master[features].apply(pd.to_numeric, errors='coerce')

final_fill_values = median_fill_values.to_dict()
final_fill_values.update({'draft_adv': 0, 'side_encoded': side_fill_value})
X_train = X_train.fillna(final_fill_values)
X_test = X_test.fillna(final_fill_values)

y_train = train_master['result'].astype(int)
y_test = test_master['result'].astype(int)

print(f"Train shape: {X_train.shape}")
print(f"Test shape: {X_test.shape}")
print(f"Missing values remaining in X_train: {int(X_train.isna().sum().sum())}")
print(f"Missing values remaining in X_test: {int(X_test.isna().sum().sum())}")

print("\nTarget distribution in train and test:")
target_distribution = pd.DataFrame({
    'train_count': y_train.value_counts().sort_index(),
    'test_count': y_test.value_counts().sort_index()
}).fillna(0).astype(int)
display(target_distribution)

print(f"Final feature list: {features}")
```

```python
# 9. SCALING
# Scaling is mainly useful for Logistic Regression, SVM, and MLP.
# X_train and X_test are also kept so tree-based models can use unscaled data if desired.
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print(f"Scaled training matrix shape: {X_train_scaled.shape}")
print(f"Scaled testing matrix shape: {X_test_scaled.shape}")
```

```python
# 10. MODELING SECTION
print("--- Baseline Models ---")
baseline_models = {
    "Logistic Regression": LogisticRegression(),
    "Support Vector Machine": SVC(kernel='linear', probability=True, random_state=42)
}

for name, model in baseline_models.items():
    model.fit(X_train_scaled, y_train)
    y_pred = model.predict(X_test_scaled)
    print(f"{name:25} | Accuracy: {accuracy_score(y_test, y_pred):.3f} | F1: {f1_score(y_test, y_pred):.3f}")

print("\n--- Advanced Models (Untuned) ---")
advanced_models = {
    "Gradient Boosting": GradientBoostingClassifier(n_estimators=100, learning_rate=0.1, max_depth=4, random_state=42),
    "AdaBoost": AdaBoostClassifier(n_estimators=100, learning_rate=1.0, random_state=42),
    "Neural Network (MLP)": MLPClassifier(hidden_layer_sizes=(64, 32), max_iter=1000, random_state=42)
}

# The original prototype uses the scaled matrices here for consistency.
for name, model in advanced_models.items():
    model.fit(X_train_scaled, y_train)
    y_pred = model.predict(X_test_scaled)
    print(f"{name:25} | Accuracy: {accuracy_score(y_test, y_pred):.3f} | F1: {f1_score(y_test, y_pred):.3f}")

print("\n--- Hyperparameter Tuning (Random Forest) ---")
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [6, 8, 10, None],
    'min_samples_split': [2, 5]
}

grid_search = GridSearchCV(RandomForestClassifier(random_state=42), param_grid, cv=5, n_jobs=1)
grid_search.fit(X_train_scaled, y_train)
best_rf = grid_search.best_estimator_

tuned_y_pred = best_rf.predict(X_test_scaled)
print(f"Tuned Random Forest       | Accuracy: {accuracy_score(y_test, tuned_y_pred):.3f} | F1: {f1_score(y_test, tuned_y_pred):.3f}")

# Export pipeline assets for the small Flask app / later reuse.
os.makedirs('models', exist_ok=True)
joblib.dump(best_rf, 'models/rf_model.pkl')
joblib.dump(scaler, 'models/scaler.pkl')
joblib.dump(train_champ_win_rates, 'models/champ_win_rates.pkl')
print("\nSuccess! Exported rf_model.pkl, scaler.pkl, and champ_win_rates.pkl to /models")
```
